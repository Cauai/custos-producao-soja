# Exportar Gold para Parquet

Le `gold.fato_custo` do Postgres e salva em `dados/gold_fato_custo.parquet` —
o snapshot que o dashboard usa no deploy.

Rode com o banco no ar e a Gold construida (notebook 04).

```python
import sys
from pathlib import Path

import pandas as pd

sys.path.append(str(Path.cwd().parent / "pipeline"))
from db import get_engine

engine = get_engine()

fato = pd.read_sql("SELECT * FROM gold.fato_custo", engine)
print(f"Linhas lidas do banco: {len(fato):,}")
```

```python
destino = Path("..") / "dados" / "gold_fato_custo.parquet"
destino.parent.mkdir(parents=True, exist_ok=True)

fato.to_parquet(destino, index=False, compression="snappy")

conferido = pd.read_parquet(destino)
print(f"Linhas no Parquet: {len(conferido):,}")
print("OK" if len(fato) == len(conferido) else "ATENCAO: contagens diferentes")
print(f"Arquivo: {destino.resolve()}")
```
